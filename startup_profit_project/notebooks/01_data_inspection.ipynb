# 01 - Data Inspection
**Dataset:** 50_Startups_dataset.csv

Goal: load the raw data and understand its structure, types, and basic quality before any analysis.

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
df = pd.read_csv('../data/50_Startups_dataset.csv', index_col=0)
df.head(10)
```

## Shape and columns

```python
print("Rows, Columns:", df.shape)
print("\nColumn names:", list(df.columns))
df.info()
```

## Data types

```python
df.dtypes
```

## Missing values

```python
missing = df.isna().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
```

## Duplicate rows

```python
print("Duplicate rows:", df.duplicated().sum())
```

## Categorical column: State

```python
print(df['State'].value_counts())
print("\nUnique states:", df['State'].unique())
```

## Numeric summary

```python
df.describe().T
```

## Basic sanity checks (negative or zero values)

```python
num_cols = ['R&D Spend', 'Administration', 'Marketing Spend', 'Profit']
for c in num_cols:
    n_zero = (df[c] == 0).sum()
    n_neg = (df[c] < 0).sum()
    print(f"{c}: zeros={n_zero}, negatives={n_neg}, min={df[c].min()}, max={df[c].max()}")
```

## Inspection notes

- 50 rows, 5 columns: `R&D Spend`, `Administration`, `Marketing Spend`, `State` (categorical), `Profit` (target).
- No missing values, no duplicate rows.
- `State` has 3 categories: New York, California, Florida — roughly balanced.
- Some `R&D Spend` and `Marketing Spend` values are effectively 0 (0.1), which are valid startups with near-zero spend in that category, not missing-data artifacts.
- All numeric columns are non-negative, consistent with spend/profit figures.
- Dataset is clean enough to proceed directly to EDA — no imputation or de-duplication needed.
